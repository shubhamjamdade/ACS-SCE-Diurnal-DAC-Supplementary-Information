```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the Excel file
file_path = 'Comparison among the region specific sites.xlsx'
sheet_name = 'Comparison among Michigan 2013'
df = pd.read_excel(file_path, sheet_name=sheet_name)

# Define the columns containing x and y values
x_column = 'A'

y_columns_list = [['B'], ['D'], ['C']]

# Define specific rows for plotting
start_row = 32
end_row = 40

# Create a single set of axes for all subplots
fig, ax = plt.subplots(figsize=(12, 8))

# Define legends
# legends = ['ELV:175, TWH:2.7, TH:177.7', 'ELV:180, TWH:2, TH:182', 'ELV:230, TWH:32, TH:262']

legends = ['US-WPT', 'US-CRT', 'US-Oho']

# Loop through the list of y_columns_list
for i, (y_columns, legend) in enumerate(zip(y_columns_list, legends)):

    # Extract x values
    x_values = df.loc[start_row:end_row, x_column]

    # Extract y values from multiple columns for the specified rows
    y_values = df.loc[start_row:end_row, y_columns]

    # Plot x and y values
    for col in y_values.columns:
        ax.plot(x_values, y_values[col], label=legend)

# Add labels and title to the plot
ax.set_xlabel('Hour of the Day', color='black', fontsize=20)
ax.set_ylabel('Ambient [CO$_2$] (ppm)', color='black', fontsize=20)
ax.set_title('Diurnal CO$_2$ variations at Michigan sites in July 2013', color='black', fontsize=18)
ax.legend(loc='upper center', fontsize=18, edgecolor='black', framealpha=1)

# Specify x-ticks and y-ticks
ax.tick_params(axis='x', colors='black', labelsize=20)
ax.tick_params(axis='y', colors='black', labelsize=20)
ax.set_xticks([0, 3, 6, 9, 12, 15, 18, 21])  # Pass a list of specific x-ticks
ax.set_yticks([360, 365, 370, 375, 380, 385, 390, 395, 400, 405, 410, 415, 420, 425])  # You can also specify y-ticks here

# Remove gridlines
ax.grid(False)

# Show plot
plt.tight_layout()
plt.show()
```
