```python
import pandas as pd

# Replace 'your_excel_file.xlsx' with the path to your Excel file
excel_file = 'AMF_CA-TP1_BASE_HH_3-5.xlsx'

# Replace 'sheet_name' with the name of the sheet you want to read
sheet_name = 'AMF_CA-TP1_BASE_HH_3-5'  # Change 'Sheet1' to the name of your sheet if it's different


# Read data from the specified sheet into a DataFrame
df = pd.read_excel(excel_file, sheet_name=sheet_name, skiprows=2)

# Read the columns in the DataFrame
columns = df.columns
```

```python
df.head(10)
```

```python
# Specify the particular value you're interested in

year = 2013

month = 7

min_start = 0

# Create a boolean mask to filter rows where column 'Min Start' has the target value
mask_start_time = df['Min Start'] == min_start

# Apply the mask to the DataFrame to filter rows
filtered_start_time_df = df[mask_start_time]

# Create a boolean mask to filter rows where column 'Year' has the target value
mask_year = filtered_start_time_df['Year'] == year

# Apply the mask to the DataFrame to filter rows
filtered_year_df = filtered_start_time_df[mask_year]

# Create a boolean mask to filter rows where column 'Month' has the target value
mask_month = filtered_year_df['Month'] == month

# Apply the mask to the DataFrame to filter rows
filtered_month_df = filtered_year_df[mask_month]

filtered_month_df.head(10)
```

```python
# Assuming 'target_column' is the column for which you want to calculate the average
P1 = 'CO2'

# Define a list of columns (P1)
columns = [P1]

# Create a dictionary to hold the averages for each column
averages_dict = {col: [] for col in columns}

num_rows = 2

# Iterate over each row in the DataFrame
for i in range(len(filtered_month_df)):
    # Iterate over each column (P1)
    for col in columns:
        # Calculate the average of the next three rows for the specific column
        if i <= len(filtered_month_df) - num_rows:
            # Get the next few rows for the current column
            next_three_rows = filtered_month_df.iloc[i:i + num_rows + 1][col]
            
            if (next_three_rows == -9999).all():
                average_next_three = -9999
            else:
                # Filter out invalid values (-9999) and calculate the mean
                valid_values = next_three_rows[next_three_rows != -9999]
                average_next_three = valid_values.mean()
        else:
            # Handle the case where there are fewer than three rows remaining
            average_next_three = None
        
        # Append the calculated average to the appropriate list in the dictionary
        averages_dict[col].append(average_next_three)

# Retrieve the averages for each column
Avg_P1 = averages_dict[P1]

```

```python
# Add the list of averages as a new column to the DataFrame
filtered_month_df = filtered_month_df.copy()
filtered_month_df['Avg_P1'] = Avg_P1



# Display the DataFrame with the new column
#print(filtered_month_df)


HOD = [0, 3, 6, 9, 12, 15, 18, 21]

# Define the columns you want to extract data from
columns_to_extract = ['Avg_P1']

# Dictionary to store extracted DataFrames for each hour
extracted_data_dict = {}
monthly_average_dict = {}

for hour in HOD:
    # Convert hour to string
    Hour_of_Day = hour
    
    # Create a mask for the current hour
    mask_HOD = filtered_month_df['Hour of the day'] == Hour_of_Day
    
    # Apply the mask to filter rows for the current hour
    filtered_hour_df = filtered_month_df[mask_HOD]
    
    # Extract data from corresponding columns for the current hour
    extracted_data = filtered_hour_df[columns_to_extract]
    
    # Store the extracted data in the dictionary
    extracted_data_dict[Hour_of_Day] = extracted_data
    
    #print(extracted_data_dict[Hour_of_Day])
    
    #monthly_average_dict[Hour_of_Day] = extracted_data_dict[Hour_of_Day].mean()
    
    # Create an empty dictionary to store the monthly average for each hour of day
    monthly_average_dict = {}

    # Iterate over each key (hour_of_day) in extracted_data_dict
    for hour_of_day, data_frame in extracted_data_dict.items():
        # Exclude rows where the values are equal to -9999
        valid_data = data_frame.replace(-9999, pd.NA)  # Replace -9999 with NA (missing value)

        # Calculate the mean of each column, excluding NA values
        mean_values = valid_data.mean(skipna=True)

        # Store the mean values in the monthly_average_dict with the key being the hour_of_day
        monthly_average_dict[hour_of_day] = mean_values

# for column in columns_to_extract:
#     print(monthly_average_dict[0][column])
#     print(monthly_average_dict[3][column])
#     print(monthly_average_dict[6][column])
#     print(monthly_average_dict[9][column])
#     print(monthly_average_dict[12][column])
#     print(monthly_average_dict[15][column])
#     print(monthly_average_dict[18][column])
#     print(monthly_average_dict[21][column]) 
#     print(column+'\n')
```

```python
import matplotlib.pyplot as plt

# List of columns to extract and plot
columns_to_extract = ['Avg_P1']

# Define the hours of the day to plot against
hours_of_day = [0, 3, 6, 9, 12, 15, 18, 21]

# Create a figure and axes for the plots
fig, ax = plt.subplots(figsize=(10, 6))

# Plot each column from monthly_average_dict for the specified hours
for column in columns_to_extract:
    # Extract values from monthly_average_dict for the specified hours
    values = [monthly_average_dict[hour][column] for hour in hours_of_day if hour in monthly_average_dict]
    
    # Plot the values against the hours of day
    ax.plot(hours_of_day, values, marker='o', label=column)

# Add labels and title
ax.set_xlabel('Hour of Day')
ax.set_ylabel('Ambient [CO2] (ppm)')
ax.set_title('Meter scale CO2 variations at US-WCr site in July 2013')

# Add a legend to distinguish the lines
ax.legend()

# Display the plot
plt.show()
```
